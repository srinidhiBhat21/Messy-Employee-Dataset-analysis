```python
import pandas as pd
import os
from warnings import filterwarnings
filterwarnings('ignore')
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("desolution01/messy-employee-dataset")

print("Path to dataset files:", path)
```

```python
print(os.listdir(path))
```

```python
df = pd.read_csv(os.path.join(path, 'Messy_Employee_dataset.csv'))
```

## We can see that Join_Date column is in object datatype, convert it to datetime format

```python
df['Join_Date'] = pd.to_datetime(df['Join_Date'],errors='coerce')
```

```python
df.dtypes
```

## Removing WhiteSpaces

```python
df['Department_Region'] = df['Department_Region'].str.strip()
```

```python
df.head()
```

```python
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.strip().str.title()
```

```python
df.head(2)
```

## Check for duplicates

```python
df.duplicated().sum()
```

## Handle Missing Values

```python
df.isnull().sum()
```

### both are numerical columns so filling missing values with median

```python
df.fillna(df.select_dtypes(include='number').median(), inplace=True)
```

```python
df.isnull().sum()
```

```python
# Dataset dimensions
print("Shape:", df.shape)

# Data types
print("\nData Types:")
print(df.dtypes)

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())

# Dataset information
df.info()

# First five rows
df.head()
```
